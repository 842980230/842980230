# **Projecto: Entendendo as Correlações**

A análise de correlação é uma técnica estatística amplamente utilizada para medir e interpretar as relações entre duas variáveis. Três métodos principais são frequentemente aplicados: **Correlação de Pearson**, **Correlação de Kendall** e **Correlação de Spearman**. Embora todos quantifiquem a relação entre variáveis, cada um tem suas particularidades e aplicações específicas.

image.png

## **Introdução às correlações**

### **Correlação de Pearson**

A Correlação de Pearson é a métrica mais comum e mede a força e a direção da relação linear entre duas variáveis quantitativas. Seu valor varia entre -1 e 1:

- 1 indica uma correlação linear perfeita positiva.
- -1 indica uma correlação linear perfeita negativa.
- 0 indica ausência de correlação linear.

A fórmula da correlação de Pearson é:

image.png

**Aplicações**:
- É apropriada quando ambas as variáveis são contínuas e seguem uma distribuição aproximadamente normal.
- Sensível a outliers, que podem distorcer os resultados.

**Exemplo:**

Considere a relação entre a altura e o peso de um grupo de pessoas. Uma correlação de Pearson alta sugeriria que pessoas mais altas tendem a pesar mais.

### **Correlação de Kendall**

A Correlação de Kendall mede a associação baseada no número de concordâncias e discordâncias entre pares ordenados de observações. Seu coeficiente, chamado de tau de Kendall, varia entre -1 e 1, com interpretações similares às de Pearson.

image.png

**Aplicações**:
- Ideal para variáveis ordinais ou quando os dados contêm muitos empates.
- Menos sensível a outliers, sendo útil em casos com distribuições assimétricas.

**Exemplo**:

Imagine classificar o desempenho de alunos em duas disciplinas. O coeficiente de Kendall indicaria a consistência entre os rankings.

### **Escala dos Coeficientes de Correlação**

Os coeficientes de correlação, independentemente do método utilizado (Pearson, Spearman ou Kendall), variam entre -1 e 1. Essa escala é crucial para interpretar a força e a direção da relação entre duas variáveis. Abaixo está uma explicação detalhada sobre o que significam os valores em diferentes intervalos:

image.png

**1.Valores Positivos (+):**

- **0 a +0.20**: Correlação muito fraca ou insignificante
Indica que há pouca ou nenhuma associação entre as variáveis. Por exemplo, um coeficiente de +0.10 sugere uma relação quase inexistente.

- **+0.20 a +0.50**: Correlação fraca
Há uma relação positiva, mas ainda não é forte. Uma variável tende a aumentar ligeiramente conforme a outra aumenta.

- **+0.50 a +0.70**: Correlação moderada
Sugere uma associação positiva clara, mas não perfeita. À medida que uma variável aumenta, a outra também tende a aumentar de maneira consistente.

- **+0.70 a +1.00**: Correlação forte ou muito forte
Indica uma relação positiva significativa. Valores próximos de +1.00 sugerem que as variáveis se movem em perfeita harmonia.

**2. Valores Negativos (-):**

- **0 a -0.20:** Correlação muito fraca ou insignificante
Assim como os valores positivos, aqui há pouca ou nenhuma associação entre as variáveis, mas a relação é inversa.

- **-0.20 a -0.50**: Correlação fraca
Há uma relação inversa, mas não muito acentuada. Quando uma variável aumenta, a outra tende a diminuir ligeiramente.

- **-0.50 a -0.70**: Correlação moderada
Mostra uma associação inversa clara. Um aumento em uma variável geralmente está associado a uma diminuição significativa na outra.

- **-0.70 a -1.00**: Correlação forte ou muito forte
Representa uma relação inversa significativa. Valores próximos de -1.00 sugerem que, quando uma variável aumenta, a outra diminui de forma praticamente perfeita.

**3. Zero (0)**:

- Um coeficiente de 0 indica nenhuma correlação linear (para Pearson) ou monotônica (para Spearman e Kendall). Nesse caso, não há nenhuma relação discernível entre as variáveis.

### **Correlações Não Implicam Causalidade**

Um dos conceitos mais importantes ao interpretar correlações é entender que correlação não implica causalidade. Isso significa que, mesmo que duas variáveis estejam correlacionadas, isso não necessariamente indica que uma variável causa a outra. Este é um ponto fundamental em análises estatísticas, para evitar conclusões erradas ou simplistas sobre os dados.

**Exemplos de Correlações Sem Causalidade**

## **Aplicação com Python**

### **Aplicação**

Vamos criar um exemplo prático em Python utilizando a biblioteca pandas para calcular os coeficientes de correlação de Pearson, Kendall e Spearman.

```python
# Bibliotecas
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
# Carregar o dataset Iris
df = sns.load_dataset('iris')

# Exibir as primeiras linhas do dataset
print('Visualizando o dataset Iris:')
df.head()
```

```python
# Selecionar apenas as colunas numéricas para cálculo de correlações
numerical_columns = df.select_dtypes(include=['float64']).columns
```

```python
# Correlação de Pearson
print('\nCorrelação de Pearson:')
pearson_corr = df[numerical_columns].corr(method='pearson')
print(pearson_corr)

# Correlação de Spearman
print('\nCorrelação de Spearman:')
spearman_corr = df[numerical_columns].corr(method='spearman')
print(spearman_corr)

# Correlação de Kendall
print('\nCorrelação de Kendall:')
kendall_corr = df[numerical_columns].corr(method='kendall')
print(kendall_corr)
```

### **Insights Comparando as Correlações**
<br>

**1. Relação mais forte entre petal_length e petal_width:**
- Em todas as três correlações (Pearson: 0.962, Spearman: 0.937, Kendall: 0.806), a relação entre petal_length e petal_width é a mais forte no dataset. Isso indica que essas variáveis têm uma relação linear e monotônica muito consistente. A menor diferença entre os coeficientes demonstra que as duas variáveis estão fortemente correlacionadas, independentemente do método utilizado.

<br>

**2. Variação no impacto de relações fracas:**

A correlação entre sepal_length e sepal_width é fraca em todos os métodos, mas varia:
- Pearson: -0.117 (praticamente sem relação linear).
- Spearman: -0.167 (uma leve relação monotônica inversa).
- Kendall: -0.077 (ainda mais fraca).

Isso mostra que, enquanto a relação linear é praticamente inexistente (Pearson), métodos baseados em rankings como Spearman e Kendall capturam uma leve tendência monotônica.

<br>

**3. Robustez a relações não lineares e monotônicas:**

A correlação de Spearman e Kendall são mais robustas para capturar relações monotônicas, enquanto Pearson é sensível apenas a relações lineares. Por exemplo, a relação entre sepal_length e petal_length é alta nos três métodos:
- Pearson: 0.872
- Spearman: 0.882
- Kendall: 0.719

Isso sugere que, embora exista uma relação linear forte entre essas variáveis, métodos baseados em rankings (Spearman e Kendall) confirmam que a tendência monotônica permanece consistente, mesmo que não seja estritamente linear.


Esses insights ajudam a escolher o método mais apropriado dependendo do tipo de relação que se deseja investigar. Pearson é útil para relações lineares, enquanto Spearman e Kendall são mais adequados para dados com tendências monotônicas ou não lineares.

## **Projeto Mercado Financeiro**

**Projeto**

Neste projeto, vamos analisar a relação entre o desempenho do índice B3 (IBOV) e a cotação do dólar (USD/BRL) durante o ano de 2024, com foco especial no mês de dezembro. Para isso, utilizaremos diferentes métricas de correlação — Pearson, Spearman e Kendall — para compreender como essas duas variáveis interagem.

A análise começa com a **coleta de dados históricos do IBOV** e do **dólar por meio da API do Yahoo Finance**. Em seguida, os dados são organizados e filtrados para garantir a consistência e qualidade das informações. Posteriormente, aplicaremos as três abordagens estatísticas para calcular as correlações e explorar a relação entre as variações do mercado acionário e a oscilação cambial.

```python
# Importando os Frameworks Necessários
import pandas as pd
import yfinance as yf
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Baixar os dados do IBOV e do dólar (USDBRL) usando a API do Yahoo Finance
ibov = yf.download( tickers='^BVSP', start='2024-12-01', end='2024-12-31')
usdbrl = yf.download(tickers='USDBRL=X', start='2024-12-01', end='2024-12-31')
```

```python
# Ajustar as colunas para um único nível, caso venham como MultiIndex
ibov.columns = [col[0] if isinstance(col, tuple) else col for col in ibov.columns]
usdbrl.columns = [col[0] if isinstance(col, tuple) else col for col in usdbrl.columns]
```

```python
# Verificando o dataSet
ibov.head()
```

```python
# Filtrar apenas os preços de fechamento
ibov_close = ibov['Close'].rename('IBOV')
usdbrl_close = usdbrl['Close'].rename('USD/BRL')
```

```python
# Unir os dados em um único DataFrame
data = pd.concat([ibov_close, usdbrl_close], axis=1)
data.head()
```

```python
# Calcular as correlações
pearson_corr = data.corr(method='pearson')
spearman_corr = data.corr(method='spearman')
kendall_corr = data.corr(method='kendall')

# Exibir as correlações
print('Correlação de Pearson:')
print(pearson_corr)

print('\nCorrelação de Spearman:')
print(spearman_corr)

print('\nCorrelação de Kendall:')
print(kendall_corr)
```

Em todas as métricas de correlação (Pearson: -0.50, Spearman: -0.55, Kendall: -0.33), observa-se uma relação negativa entre o índice B3 (IBOV) e o dólar (USD/BRL).

### **Interpretação**

**Força da Relação**:
- **Pearson (-0.50)**: Indica uma relação linear moderada. O movimento de preços do IBOV e do dólar segue uma tendência oposta, mas não completamente linear.
- **Spearman (-0.55)**: Indica uma relação monotônica mais forte, sugerindo que quando uma variável sobe, a outra consistentemente desce, mesmo que a relação não seja perfeitamente linear.
- **Kendall (-0.33)**: Confirma a consistência da relação monotônica, mas em um nível menos forte devido à abordagem baseada em pares de dados.


**Mercado em Dezembro**:
- Dezembro costuma ser um mês volátil para os mercados devido à sazonalidade, ajustes de portfólio e eventos econômicos globais.

**Essa correlação negativa pode indicar que**:
- O mercado de ações foi impulsionado por fatores locais, enquanto o dólar se desvalorizou devido a menor percepção de risco.
- Inversamente, uma desvalorização do dólar pode ter incentivado investidores a buscar ativos na bolsa.
